# 4.	Modelado Predictivo:

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor  # Para un modelo de regresión
from sklearn.metrics import mean_squared_error
```

```python
x= X_numero[['Inches', 'Ram', 'Weight', 'ScreenW','ScreenH','CPU_freq','PrimaryStorage']]
y= y_numero[['Price_euros']]
# Dividir los datos en conjunto de entrenamiento y conjunto de prueba (80% entrenamiento, 20% prueba)
X_train, X_test, y_train, y_test = train_test_split(X_numero, y, test_size=0.2, random_state=42)
# mejorar rendimiento del modelo
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Creación del modelo Random Forest para regresión
model = RandomForestRegressor(n_estimators=100, random_state=42)

# Entrenamos el modelo con los datos de entrenamiento
model.fit(X_train, y_train)
```

# 5.	Evaluación del Modelo:

```python
# Hacer predicciones sobre el conjunto de prueba
y_pred = model.predict(X_test)

# Calculando el error cuadrático medio (MSE) y el coeficiente de determinación R^2
mse = mean_squared_error(y_test, y_pred)
r2 = model.score(X_test, y_test)

print(f'Error Cuadrático Medio (MSE): {mse}')
print(f'Coeficiente de Determinación (R²): {r2}')
```

```python
# Calcular los errores de predicción (residuales)
errores = y_test['Price_euros'] - y_pred

# Crear un histograma de los errores
plt.figure(figsize=(8, 6))
sns.histplot(errores, bins=20, kde=True, color='purple')
plt.xlabel('Error de Predicción')
plt.title('Distribución de los Errores de Predicción')
plt.show()
```

El histograma nos permite ver si el modelo tiende a subestimar o sobreestimar las ventas.
Una distribución centrada en 0 nos indica que el modelo predice bien en promedio.
