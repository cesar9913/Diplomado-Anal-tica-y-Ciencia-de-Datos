```python
import numpy as np
from scipy.misc import derivative

def g(x):
  return np.sin(x)

x = 0.0
dx = 1e-6
derivada = derivative(g, x, dx)
print(f"La derivada de g(x) en x = {x} es {derivada}")
```

```python
import numpy as np
from scipy.misc import derivative
import matplotlib.pyplot as plt

def g(x):
  return x**3 - 6*x**2 + 11*x -6

valores = np.linspace(0,3,100)
derivados = [ derivative(g, x, dx = 1e-6) for x in valores]

plt.plot(valores, derivados)
plt.show()
```

Aplicación Práctica de NumPy: Resolviendo un Problema

Crear el array de visitantes (puedes usar valores aleatorios):

```python
visitantes = np.random.randint(100, 1000, (7, 10))
print("Visitantes por día y atracción:")
print(visitantes)
```

Calcula el total de visitantes por día (suma de cada fila):

```python
total_por_dia = visitantes.sum(axis=1)
print("Total de visitantes por día:", total_por_dia)
```

Encuentra el día con más visitantes y el día con menos visitantes:

```python
dia_mas_visitantes = total_por_dia.argmax()
dia_menos_visitantes = total_por_dia.argmin()
print("Día con más visitantes:", dia_mas_visitantes)
print("Día con menos visitantes:", dia_menos_visitantes)
```

Calcula el promedio de visitantes por atracción en los 7 días:

```python
promedio_por_atraccion = visitantes.mean(axis=0)
print("Promedio de visitantes por atracción:", promedio_por_atraccion)
```

Filtra las atracciones que han tenido más de 800 visitantes en algún día:

```python
atracciones_muy_populares = visitantes[visitantes > 800]
print("Atracciones con más de 800 visitantes en algún día:", atracciones_muy_populares)
```
