```python
class Vehiculo:
    def __init__(self, marca, modelo, año):
        self.marca = marca
        self.modelo = modelo
        self.año = año

    def descripcion(self):
        return f"{self.marca} {self.modelo} del {self.año}"

class Auto(Vehiculo):
    def __init__(self, marca, modelo, año, tipo_combustible):
        super().__init__(marca, modelo, año)
        self.tipo_combustible = tipo_combustible

    def descripcion_completa(self):
        return f"{self.descripcion()} que usa {self.tipo_combustible}"

mi_auto = Auto("Toyota", "Corolla", 2020, "gasolina")
print(mi_auto.descripcion_completa())
```

Ejercicio El Ahorcado con Funciones

Ejercicio:

    Crea una función llamada jugar_ahorcado que implemente el clásico juego del ahorcado.
    Utiliza funciones para manejar la lógica del juego, como elegir una palabra, verificar letras, y mantener el marcador.

```python
import random

def elegir_palabra():
    palabras = ['python', 'jugar', 'clase', 'funcion']
    return random.choice(palabras)

def jugar_ahorcado():
    palabra = elegir_palabra()
    intentos = 6
    letras_adivinadas = []

    while intentos > 0:
        letra = input("Adivina una letra: ")
        if letra in palabra:
            letras_adivinadas.append(letra)
            print(f"¡Correcto! Las letras adivinadas son: {letras_adivinadas}")
            #
            break
        else:
            intentos -= 1
            print(f"¡Incorrecto! Te quedan {intentos} intentos.")

        if set(letras_adivinadas) == set(palabra):
            print(f"¡Ganaste! La palabra era {palabra}")
            break
    else:
        print(f"Perdiste. La palabra era {palabra}")
#Mandar llamar a la funcion jugar_ahorcado
jugar_ahorcado()
```
