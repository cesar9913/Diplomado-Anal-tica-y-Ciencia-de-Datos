```python
import matplotlib.pyplot as plt
import numpy as np

# Datos de ejemplo
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Crear gráfico de líneas
plt.figure(figsize=(8,5))
plt.plot(x, y1, label='Seno')
plt.plot(x, y2, label='Coseno')

# Agregar título y etiquetas
plt.title("Gráfico de Líneas de Seno y Coseno")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()

# Mostrar gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt

# Datos de ejemplo
x = np.random.rand(50);print(type(x));print(x)
y = np.random.rand(50);print(y)

# Crear gráfico de dispersión
plt.scatter(x, y, color='b')

# Agregar etiquetas y título
plt.title("Gráfico de Dispersión")
plt.xlabel("Variable X")
plt.ylabel("Variable Y")

# Mostrar gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt

# Datos de ejemplo (distribución normal)
data = np.random.randn(1000)

# Crear histograma
plt.hist(data, bins=30, color='blue', alpha=0.7)

# Agregar título y etiquetas
plt.title("Histograma")
plt.xlabel("Valores")
plt.ylabel("Frecuencia")

# Mostrar gráfico
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar dataset de ejemplo
tips = sns.load_dataset("tips");print(tips)
# Crear gráfico de dispersión con Seaborn
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", style="time")

# Mostrar gráfico
plt.title("Relación entre Total de la Cuenta y Propina")
plt.show()

```

```python
# Cargar el dataset del Titanic desde Seaborn
titanic_df = sns.load_dataset('titanic')

# Mostrar las primeras filas del DataFrame para conocer las variables
print(titanic_df.head())

# Crear el gráfico de conteo
sns.countplot(x='survived', data=titanic_df)

# Mostrar el gráfico
plt.title('Supervivencia en el Titanic')
plt.show()
```

4. REPRESENTAR HISTOGRAMAS CON MATPLOTLIB

```python
criptodatos = pd.read_csv('crypto_daily_returns.csv')
```

```python
media = round(criptodatos['BTC'].mean(),2) # media de la distibucion
sigma = round(criptodatos['BTC'].std(),2) # desviacion estandar
num_bins = 40
plt.figure(figsize=(20,15))
criptodatos['BTC'].plot.hist(bins = num_bins, color = 'b', alpha = 0.5, grid = True, figsize=(12,7))

plt.grid()
plt.ylabel('Probabilidad')
plt.title('Histograma: medua = ' + str(media) + ', desviacion estandar = ' + str(sigma))
plt.show()
```
