```python
# Ejemplo práctico:
import matplotlib.pyplot as plt
import numpy as np

# Datos de ejemplo
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Crear gráfico de líneas
plt.figure(figsize=(8,5))
plt.plot(x, y1, label='Seno')
plt.plot(x, y2, label='Coseno')

# Agregar título y etiquetas
plt.title("Gráfico de Líneas de Seno y Coseno")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()

# Mostrar gráfico
plt.show()
```

```python
# Teoría:
# Un gráfico de dispersión (scatter plot) se usa para mostrar la relación entre
# dos variables numéricas. Es útil para detectar patrones o correlaciones.

# Ejemplo práctico:
import matplotlib.pyplot as plt

# Datos de ejemplo
x = np.random.rand(50)
y = np.random.rand(50)

# Crear gráfico de dispersión
plt.scatter(x, y, color='r')

# Agregar etiquetas y título
plt.title("Gráfico de Dispersión")
plt.xlabel("Variable X")
plt.ylabel("Variable Y")

# Mostrar gráfico
plt.show()
```

```python
# 1.3. Representar Histogramas con Matplotlib
# Teoría:
# Un histograma se utiliza para representar la distribución de una variable continua,
# dividiendo los datos en intervalos o "bins".

# Ejemplo práctico:

import matplotlib.pyplot as plt

# Datos de ejemplo (distribución normal)
data = np.random.randn(1000)

# Crear histograma
plt.hist(data, bins=30, color='blue', alpha=0.7)

# Agregar título y etiquetas
plt.title("Histograma")
plt.xlabel("Valores")
plt.ylabel("Frecuencia")

# Mostrar gráfico
plt.show()
```

```python
# ...1.6. Gráfico de Conteo con Seaborn
# Un Gráfico de Conteo o Countplot es un tipo de gráfico de barras que muestra la
# frecuencia de ocurrencia de diferentes categorías dentro de una
# variable categórica.
# Seaborn lo facilita con la función sns.countplot().
# Este gráfico es útil para visualizar cuántas veces aparecen diferentes valores
# categóricos en un conjunto de datos...

# Ejemplo práctico:
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar dataset
tips = sns.load_dataset("tips")

# Crear gráfico de conteo
sns.countplot(data=tips, x="day")

# Mostrar gráfico
plt.title("Conteo de Propinas por Día")
plt.show()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Cargar el dataset del Titanic desde Seaborn
titanic_df = sns.load_dataset('titanic')

# Mostrar las primeras filas del DataFrame para conocer las variables
print(titanic_df.head())
# Crear el gráfico de conteo

sns.countplot(y='survived', hue='class', data=titanic_df)

# Mostrar el gráfico
plt.title('Supervivencia en el Titanic')
plt.show()
# Crear un gráfico de conteo, coloreado según la clase de pasajero
sns.countplot(x='survived', hue='class', data=titanic_df)

# Mostrar el gráfico
plt.title('Supervivencia por clase en el Titanic')
plt.show()
```
