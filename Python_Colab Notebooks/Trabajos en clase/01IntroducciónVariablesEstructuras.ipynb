```python
# Conversor de temperaturas
opcion = input("¿Quieres convertir de Celsius a Fahrenheit (C) o de Fahrenheit a Celsius (F)? ")

if opcion.lower() == 'c':
    celsius = float(input("Ingresa la temperatura en grados Celsius: "))
    fahrenheit = (9/5) * celsius + 32
    print(f"{celsius}°C son {fahrenheit}°F")
elif opcion.lower() == 'f':
    fahrenheit = float(input("Ingresa la temperatura en grados Fahrenheit: "))
    celsius = (5/9) * (fahrenheit - 32)
    print(f"{fahrenheit}°F son {celsius}°C")
else:
    print("Opción no válida. Intenta de nuevo.")
```

```python
# Calculadora de IMC
peso = float(input("Ingresa tu peso en kilogramos: "))
altura = float(input("Ingresa tu altura en metros: "))

imc = peso / (altura ** 2)

print(f"Tu IMC es: {imc:.2f}")

# Clasificación del IMC
if imc < 18.5:
    print("Estás en la categoría de bajo peso.")
elif 18.5 <= imc < 25:
    print("Estás en la categoría de peso saludable.")
elif 25 <= imc < 30:
    print("Estás en la categoría de sobrepeso.")
else:
    print("Estás en la categoría de obesidad.")
```

3.1. Ejercicio Juego de Adivinanza
Descripción: Crear un juego donde la computadora elige un número aleatorio entre 1 y 100, y el jugador debe adivinarlo.

Instrucciones:

La computadora elige un número al azar entre 1 y 100.
El jugador tiene que adivinar el número, y la computadora le da pistas si el número es mayor o menor que el que se ha elegido.
El juego continúa hasta que el jugador adivine el número.


```python
import random

numero_aleatorio = random.randint(1, 100)
print(numero_aleatorio)
adivinanza = 9

print("¡Adivina el número entre 1 y 100!")

while adivinanza != numero_aleatorio:
    adivinanza = int(input("Introduce tu número: "))

    if adivinanza < numero_aleatorio:
        print("El número es mayor.")
    elif adivinanza > numero_aleatorio:
        print("El número es menor.")
    else:
        print("¡Felicidades! Has adivinado el número.")
```

3.3. Ejercicio Calculadora de Factoriales
Descripción: Crear un programa que calcule el factorial de un número usando un ciclo while.

Instrucciones:

El usuario ingresa un número entero.
El programa calcula el factorial de ese número usando un ciclo while.
Se debe asegurar que el número ingresado sea positivo.
numero = int(input("Ingresa un número entero positivo: "))

```python
numero = int(input("Ingresa un número entero positivo: "))

if numero < 0:
    print("El número debe ser positivo.")
else:
    factorial = 1
    i = 1

    while i <= numero:
        factorial *= i
        i += 1

    print(f"El factorial de {numero} es {factorial}.")
```

3.5. Ejercicio Números Primos
Descripción: Crear un programa que determine si un número ingresado por el usuario es primo o no.

Instrucciones:

El usuario ingresa un número.
El programa determina si el número es primo (solo divisible entre 1 y sí mismo) utilizando un ciclo for.


```python
numero = int(input("Ingresa un número: "))

if numero < 2:
    print(f"{numero} no es un número primo.")
else:
    es_primo = True
    for i in range(2, numero):
        if numero % i == 0:
            es_primo = False
            break

    if es_primo:
        print(f"{numero} es un número primo.")
    else:
        print(f"{numero} no es un número primo.")
```

3.6. Ejercicio Calculadora de Potencias
Descripción: Crear un programa que calcule la potencia de un número elevado a otro utilizando un ciclo for.

Instrucciones:

El usuario ingresa una base y un exponente.
El programa calcula la potencia de la base elevada al exponente utilizando un ciclo

```python
base = int(input("Ingresa la base: "))
exponente = int(input("Ingresa el exponente: "))

resultado = 1

for _ in range(exponente):
    resultado *= base

print(f"{base} elevado a {exponente} es {resultado}.")
```

3.8. Ejercicio Cálculo de Suma de Números Pares
Descripción: Crear un programa que sume todos los números pares entre 1 y un número ingresado por el usuario.

Instrucciones:

El usuario ingresa un número.
El programa suma todos los números pares desde 1 hasta el número ingresado usando un ciclo for.


```python
numero = int(input("Ingresa un número: "))
suma_pares = 0

for i in range(1, numero + 1):
    print(i)
    if i % 2 == 0:
        suma_pares += i

print(f"La suma de los números pares hasta {numero} es {suma_pares}.")
```

3.9. Ejercicio Calculadora de Promedios
Descripción: Crear un programa que calcule el promedio de una lista de números ingresados por el usuario.

Instrucciones:

El usuario ingresa una cantidad de números.
El programa calcula el promedio de esos números usando un ciclo for.

```python
n = int(input("¿Cuántos números quieres ingresar? "))
suma = 0

for i in range(n):
    numero = float(input(f"Ingresa el número {i + 1}: "))
    suma += numero

promedio = suma / n
print(f"El promedio de los números es {promedio}.")
```

3.10. Ejercicio Números Perfectos
Descripción: Crear un programa que determine si un número ingresado por el usuario es un "número perfecto". Un número perfecto es un número que es igual a la suma de sus divisores propios positivos (excluyendo el propio número).

Instrucciones:

El usuario ingresa un número.
El programa determina si es un número perfecto usando ciclos for.


```python
numero = int(input("Ingresa un número: "))
suma_divisores = 0

for i in range(1, numero):
    if numero % i == 0:
        suma_divisores += i

if suma_divisores == numero:
    print(f"{numero} es un número perfecto.")
else:
    print(f"{numero} no es un número perfecto.")
```
